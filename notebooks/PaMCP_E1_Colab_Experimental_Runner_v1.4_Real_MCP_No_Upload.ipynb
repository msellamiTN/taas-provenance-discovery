## 2. Provenance coverage gate

This cell reports missing source identity information. It does not infer or fill missing values. Exact column names are discovered from the uploaded intake file to avoid silently assuming a schema.

````python
def find_col(columns, candidates):
    norm = {c.strip().lower(): c for c in columns}
    for candidate in candidates:
        if candidate.lower() in norm:
            return norm[candidate.lower()]
    return None

columns = list(rows[0].keys())
required_groups = {
    "case_id": ["case_id"],
    "benchmark_source": ["source_record_uri", "source_benchmark"],
    "repository": ["mcp_server_repository"],
    "version": ["mcp_server_version"],
    "commit": ["mcp_server_commit"],
    "transport": ["transport"],
    "launch_command": ["launch_command"],
    "credential_requirement": ["credential_requirement"],
    "exact_tool_name": ["exact_tool_name_verified"],
    "schema_digest": ["tool_schema_sha256"],
    "artifact_digest": ["server_artifact_sha256"],
    "mapping_status": ["mapping_status"]
}
resolved = {k: find_col(columns, v) for k, v in required_groups.items()}
coverage = {}
for key, col in resolved.items():
    if col is None:
        coverage[key] = {"column_found": False, "populated": 0}
    else:
        populated = sum(bool((r.get(col) or "").strip()) for r in rows)
        coverage[key] = {"column_found": True, "populated": populated}
coverage
````

## 5. Validate manifest and artifact — no server launch

Edit `/content/pamcp_e1/execution_manifest.json` in the notebook runtime only after resolving a case's exact source identity. This validator is deliberately strict. It verifies the artifact digest and that the command directly targets the artifact.

````python
def canonical_json_sha256(obj):
    raw = json.dumps(obj, sort_keys=True, separators=(",", ":"), ensure_ascii=False).encode()
    return hashlib.sha256(raw).hexdigest()

def validate_manifest(m):
    required = ["case_id","benchmark_source_uri","server_repository","server_version",
                "server_commit","transport","artifact_path","artifact_sha256","command",
                "exact_tool_name","input_schema_sha256","credential_requirement",
                "fixture","execution_approved","approval_reference"]
    missing = [k for k in required if k not in m or m[k] in ("", None, [])]
    if missing:
        raise ValueError(f"Missing required fields: {missing}")
    if m["transport"] != "stdio":
        raise ValueError("This pilot notebook currently permits stdio only.")
    if m["credential_requirement"] != "none":
        raise ValueError("Credentialed servers are out of scope for this Colab pilot.")
    if not isinstance(m["command"], list) or not all(isinstance(x, str) for x in m["command"]):
        raise ValueError("command must be a list of strings; shell strings are forbidden.")
    if not (isinstance(m["server_commit"], str) and len(m["server_commit"]) == 40 and all(c in "0123456789abcdefABCDEF" for c in m["server_commit"])):
        raise ValueError("server_commit must be a full 40-character Git commit.")
    artifact = Path(m["artifact_path"]).resolve()
    if not artifact.is_file():
        raise ValueError(f"Artifact not found: {artifact}")
    actual = sha256_file(artifact)
    if actual.lower() != m["artifact_sha256"].lower():
        raise ValueError("Artifact SHA-256 mismatch.")
    cmd_paths = [Path(x).resolve() for x in m["command"] if x and not x.startswith("-") and ("/" in x or x.endswith(".py"))]
    if artifact not in cmd_paths:
        raise ValueError("The command does not directly reference artifact_path.")
    if not isinstance(m["fixture"], dict):
        raise ValueError("fixture must be a JSON object.")
    if m["execution_approved"] is not True:
        raise ValueError("execution_approved must be true; no launch permitted.")
    return {"valid": True, "artifact_sha256": actual, "artifact_path": str(artifact)}

# After editing the JSON file, run:
# m = json.loads(manifest_path.read_text())
# validate_manifest(m)
````

## 6A. Real MCP stdio execution

This addition installs a reviewed MCP Python SDK client and enables initialize, tools/list, and tools/call. It executes only after an operator supplies a verified manifest, pins the server artifact and tool schema, and explicitly approves the run. It does not infer source identity from tool names.

Use only credential-free, reviewed public servers and harmless fixtures. Colab is not a security sandbox.

````python
%pip -q install "mcp==1.26.0"
````

````python
import asyncio, json, hashlib, os, re, uuid
from datetime import datetime, timezone
from pathlib import Path
from mcp import ClientSession, StdioServerParameters
from mcp.client.stdio import stdio_client

def canonical_schema_hash(schema):
    raw = json.dumps(schema, sort_keys=True, separators=(",", ":"), ensure_ascii=False).encode()
    return hashlib.sha256(raw).hexdigest()

def append_jsonl(path, event):
    with open(path, "a", encoding="utf-8") as fp:
        fp.write(json.dumps(event, ensure_ascii=False, sort_keys=True) + "\\n")

def redact_value(value):
    raw = json.dumps(value, sort_keys=True, ensure_ascii=False).encode()
    return {"redacted": True, "sha256": hashlib.sha256(raw).hexdigest()}

async def execute_real_mcp_stdio(manifest):
    required = ["case_id","benchmark_source_uri","server_repository","server_version",
                "server_commit","artifact_path","artifact_sha256","command",
                "exact_tool_name","input_schema_sha256","fixture","approval_reference"]
    missing = [k for k in required if not manifest.get(k)]
    if missing: raise ValueError(f"Missing provenance fields: {missing}")
    if manifest.get("transport") != "stdio": raise ValueError("Only stdio is enabled.")
    if manifest.get("credential_requirement") != "none": raise ValueError("Credentialed server blocked.")
    if manifest.get("execution_approved") is not True: raise ValueError("Explicit execution approval required.")
    if not isinstance(manifest["command"], list) or not all(isinstance(x,str) for x in manifest["command"]):
        raise ValueError("command must be an argv list.")
    if not re.fullmatch(r"[0-9a-fA-F]{40}", manifest["server_commit"]):
        raise ValueError("Full immutable Git commit required.")
    artifact = Path(manifest["artifact_path"]).resolve()
    if not artifact.is_file(): raise ValueError("Pinned artifact not found.")
    actual = sha256_file(artifact)
    if actual.lower() != manifest["artifact_sha256"].lower(): raise ValueError("Artifact SHA-256 mismatch.")
    argv_paths = [Path(x).resolve() for x in manifest["command"] if x and ("/" in x or x.endswith(".py"))]
    if artifact not in argv_paths: raise ValueError("Command must directly target the hashed artifact.")
    run_id = str(uuid.uuid4())
    ledger = EVIDENCE / f"{run_id}.protocol.jsonl"
    started = datetime.now(timezone.utc).isoformat()
    append_jsonl(ledger, {"event":"run_started","run_id":run_id,"case_id":manifest["case_id"],
                         "timestamp_utc":started,"artifact_sha256":actual,
                         "server_commit":manifest["server_commit"],"credential_requirement":"none"})
    params = StdioServerParameters(command=manifest["command"][0], args=manifest["command"][1:],
        env={k:v for k,v in os.environ.items() if k in ("PATH","HOME","LANG","LC_ALL","PYTHONPATH")})
    result = {"protocol_execution":"FAILED","side_effects_observed":False,
              "scientific_effect_classification":"NOT_ESTABLISHED"}
    try:
        async with stdio_client(params) as (read_stream, write_stream):
            async with ClientSession(read_stream, write_stream) as session:
                init = await session.initialize()
                append_jsonl(ledger, {"event":"initialize","run_id":run_id,
                    "protocol_version":getattr(init,"protocolVersion",None),
                    "server_info":str(getattr(init,"serverInfo",None))[:500]})
                listed = await session.list_tools()
                matches = [t for t in listed.tools if t.name == manifest["exact_tool_name"]]
                if len(matches) != 1: raise ValueError("Exact tool name not advertised exactly once.")
                observed_hash = canonical_schema_hash(matches[0].inputSchema)
                append_jsonl(ledger, {"event":"tools_list","run_id":run_id,
                    "advertised_tool_count":len(listed.tools),"exact_tool_name":matches[0].name,
                    "input_schema_sha256":observed_hash})
                if observed_hash.lower() != manifest["input_schema_sha256"].lower():
                    raise ValueError("Advertised input schema hash mismatch.")
                append_jsonl(ledger, {"event":"tool_call_started","run_id":run_id,
                    "tool_name":matches[0].name,"arguments_digest":redact_value(manifest["fixture"])})
                call = await session.call_tool(matches[0].name, arguments=manifest["fixture"])
                append_jsonl(ledger, {"event":"tool_call_finished","run_id":run_id,
                    "is_error":bool(getattr(call,"isError",False)),
                    "content_blocks":len(getattr(call,"content",[]) or [])})
                result.update({"protocol_execution":"SUCCESS" if not getattr(call,"isError",False) else "TOOL_ERROR",
                    "advertised_tool_count":len(listed.tools),"tool_name":matches[0].name,
                    "observed_schema_sha256":observed_hash,"protocol_error":bool(getattr(call,"isError",False))})
    except Exception as exc:
        append_jsonl(ledger, {"event":"run_error","run_id":run_id,
            "error_type":type(exc).__name__,"error_message":str(exc)[:500]})
        result["error"] = f"{type(exc).__name__}: {str(exc)[:500]}"
    finally:
        append_jsonl(ledger, {"event":"run_finished","run_id":run_id,
            "timestamp_utc":datetime.now(timezone.utc).isoformat(),
            "side_effects_observed":False,"scientific_effect_classification":"NOT_ESTABLISHED"})
    result.update({"run_id":run_id,"ledger":str(ledger),"started_utc":started,
        "finished_utc":datetime.now(timezone.utc).isoformat(),"side_effects_observed":False,
        "scientific_effect_classification":"NOT_ESTABLISHED"})
    (EVIDENCE / f"{run_id}.run_summary.json").write_text(json.dumps(result,indent=2))
    return result

print("Real MCP stdio client installed. No server has been launched by this cell.")
````
